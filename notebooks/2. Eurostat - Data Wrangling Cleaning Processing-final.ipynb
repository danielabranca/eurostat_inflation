Changing the data types of date from str and int64 to datetime:

```python
#MANR
df_manr["date"] = pd.to_datetime(df_manr["date"], format="%Y-%m")
df_manr["homologous_inflation"] = pd.to_numeric(df_manr["homologous_inflation"])

#item weights
df_inw["date"] = pd.to_numeric(df_inw["date"]).astype("int16")
df_inw["item_weight"] = pd.to_numeric(df_inw["item_weight"])

#HPI
df_q["date"] = pd.PeriodIndex(df_q["date"], freq="Q")
df_q["house_price_growth_quarterly"] = pd.to_numeric(df_q["house_price_growth_quarterly"])
```

```python
#creating a key called 'year' on df_manr, for later merge with df_inw:
df_manr["year"] = df_manr["date"].dt.year
df_inw = df_inw.rename(columns={"date": "year"})
```

```python
#exporting wrangled dataframes:
df_manr.to_csv('hicp_manr_wrangled.csv', index = False)

df_inw.to_csv('hicp_inw_wrangled.csv', index = False)

df_q.to_csv('hpi_q_wrangled.csv', index = False)
```

#### Merging dataframes

```python
hicp = df_manr.merge(
    df_inw[["country", "category", "year", "item_weight"]],
    on=["country", "category", "year"],
    how="left",
    validate="m:1",
    indicator=True)

assert (hicp["_merge"] == "both").all()
hicp = hicp.drop(columns="_merge")
```
